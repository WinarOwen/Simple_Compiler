```python
from automata.fa.nfa import NFA
```

```python
one_third_position_NFA = NFA(
    states={'q1','q2','q3','q4'},
    input_symbols= {'0','1'},
    transitions={
        'q1' : {'0' : {'q1'} , '1' : {'q1','q2'}},
        'q2' : {'0' : {'q3'}, '1' : {'q3'}},
        'q3' : {'0' : {'q4'}, '1' : {'q4'}},
    },
    initial_state='q1',
    final_states={'q4'}
)
```

```python
one_third_position_NFA.accepts_input("000100")
```

```python
one_third_position_NFA.accepts_input("000010")
```

```python
one_third_position_NFA.show_diagram(path="automata_pics/one_third_position_NFA.png")
```

```python
multiple_of_two_or_threes_NFA = NFA(
    states={'q1','q2','q3','q4','q5','q6'},
    input_symbols={'0'},
    transitions={
        'q1': {'' :{'q2','q4'}},
        'q2': {'0' : {'q3'}},
        'q3': {'0' : {'q2'}},
        'q4': {'0' : {'q5'}},
        'q5': {'0' : {'q6'}},
        'q6': {'0' : {'q4'}}
    },
    initial_state='q1',
    final_states={'q2','q4'}
)
```

```python
multiple_of_two_or_threes_NFA.accepts_input("00")
```

```python
multiple_of_two_or_threes_NFA.accepts_input("000")
```

```python
multiple_of_two_or_threes_NFA.accepts_input("00000000")
```

```python
multiple_of_two_or_threes_NFA.accepts_input("00000")
```

```python
multiple_of_two_or_threes_NFA.show_diagram(path='automata_pics/multiple_of_two_or_three_NFA.png')
```
