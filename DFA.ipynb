```python
from automata.fa.dfa import DFA
```

```python
turnstile_dfa = DFA(
    states={'Locked','Unlocked'},
    input_symbols={'Push','Insert'},
    transitions={
        'Locked':{'Push' : 'Locked', 'Insert':'Unlocked'},
        'Unlocked':{'Push' : 'Locked', 'Insert' : 'Unlocked'}
    },
    initial_state= 'Locked',
    final_states= {'Locked'}
)
```

```python
turnstile_dfa.accepts_input(["Push"])
```

```python
turnstile_dfa.accepts_input(["Insert", "Push"])
```

```python
turnstile_dfa.accepts_input(["Insert","Push","Insert"])
```

```python
turnstile_dfa.show_diagram(path='automata_pics/turnstile_dfa.png')
```

```python
even_zeros_dfa = DFA(
    states={'q0','q1','q2'},
    input_symbols={'0','1'},
    transitions={
        'q0': {'0' : 'q1', '1' : 'q0'},
        'q1':{'0'  : 'q2', '1' : 'q1'},
        'q2':{'0' : 'q1', '1' : 'q2'},
    },
    initial_state='q0',
    final_states={'q2'}
)
```

```python
even_zeros_dfa.accepts_input("0011")
```

```python
even_zeros_dfa.accepts_input("00110101010")
```

```python
even_zeros_dfa.accepts_input("000110000")
```

```python
even_zeros_dfa.show_diagram(path="automata_pics/even_zeros_dfa.png")
```
